```python
n_meg_raw = MEG_raw.copy()
n_meg_raw.resample(sfreq=100)
```

```python
raw_sfreq = n_meg_raw.info['sfreq']
print(raw_sfreq)
```

```python
# 1. Get the raw data from MNE (shape: 328 channels, 44600 time points)
meg_data = n_meg_raw.get_data() 

# 2. Create an empty list to store the varying IMFs for each channel
all_imfs = []

# 3. Loop through your channels. 
# NOTE: EMD is computationally heavy. I highly recommend testing this 
# on just 2 or 3 channels first before running all 328!
num_channels_to_run = 3 # Change to meg_data.shape[0] to run all 328

for i in range(num_channels_to_run):
    print(f"Sifting channel {i+1}...")
    
    # Extract one 1D vector (length: 44600)
    single_channel = meg_data[i, :] 
    
    # Sift the single channel
    imf = emd.sift.sift(single_channel) 
    
    # Save the result
    all_imfs.append(imf)

print("---")
print(f"Successfully sifted {len(all_imfs)} channels.")
print(f"Shape of IMFs for Channel 1: {all_imfs[0].shape}")
print(f"Shape of IMFs for Channel 2: {all_imfs[1].shape}")
```

```python
#import matplotlib.pyplot as plt
#import emd

# Assuming 'all_imfs' is the list we created in the previous step
# Let's grab the IMFs for just the FIRST channel (index 0)
imf_single_channel = all_imfs[0] 

# 1. Plot the IMFs
# We can pass a figsize to make sure it isn't too cramped vertically
fig_imf = plt.figure(figsize=(10, 8))
emd.plotting.plot_imfs(imf_single_channel, fig=fig_imf)
plt.title("IMFs for Channel 1")

# --- Your HHT Code ---
# (Make sure IF and IA are also computed for this single channel!)
IP, IF, IA = emd.spectra.frequency_transform(imf_single_channel, raw_sfreq, 'hilbert')
freq_range = (0.1, 10, 80, 'log')
f, hht = emd.spectra.hilberthuang(IF, IA, freq_range, sum_time=False)

# 2. Plot the Hilbert-Huang Transform
fig_hht = plt.figure(figsize=(10, 6))
emd.plotting.plot_hilberthuang(hht, time_vect, f,
                               time_lims=(2, 4), freq_lims=(0.1, 15),
                               fig=fig_hht, log_y=True)

# 3. SHOW THE PLOTS (This is what makes the windows actually appear!)
plt.show()
```

```python
#import numpy as np
#from scipy.io import wavfile
#import emd
#import matplotlib.pyplot as plt

# 1. Load the .wav file
# wavfile.read returns the sample rate (e.g., 16000 or 44100) and the data array
sample_rate, audio_data = wavfile.read(PREDICTOR_audio)

# 2. Convert to Mono (if the audio is stereo)
# Stereo audio has shape (n_samples, 2). EMD needs a 1D array.
if audio_data.ndim > 1:
    print("Converting stereo to mono...")
    # Take just the left channel (or you could average them: np.mean(audio_data, axis=1))
    audio_data = audio_data[:, 0]

# 3. Normalize the audio and convert to float
# Audio is often read as int16. EMD requires floats, and normalizing to [-1, 1] is best practice.
audio_data = audio_data.astype(float)
audio_data = audio_data / np.max(np.abs(audio_data))

"""
# OLD CODES
# 4. CRITICAL: Take a short snippet for testing
# Audio sample rates are huge. 1 second of 44.1kHz audio = 44,100 samples.
# Sifting the whole file at once will take a massive amount of time. 
# Let's grab just the first 0.5 seconds to test.
snippet_length = int(sample_rate * 0.5) 
audio_snippet = audio_data[:snippet_length]

# 5. Run the Sift
print(f"Sifting {len(audio_snippet)} samples...")
imfs = emd.sift.sift(audio_snippet)

print(f"Extraction complete! Found {imfs.shape[1]} IMFs.")

"""
# 4. Grab a 0.5-second snippet from the MIDDLE of the audio
start_time_sec = 5.0  # Start 5 seconds in (change this to where someone is speaking)
duration_sec = 0.5    # Grab half a second

start_idx = int(start_time_sec * sample_rate)
end_idx = int((start_time_sec + duration_sec) * sample_rate)

audio_snippet = audio_data[start_idx:end_idx]

# --- THE SAFETY CHECK ---
# If the standard deviation is 0, the audio is completely flat.
if np.std(audio_snippet) == 0:
    raise ValueError("The selected audio snippet is completely silent! Change the start_time_sec.")

# 5. Run the Sift (This will now work properly)
print(f"Sifting {len(audio_snippet)} samples...")
imfs = emd.sift.sift(audio_snippet)

print(f"Extraction complete! Found {imfs.shape[1]} IMFs.")

# 6. Plot the results
fig = plt.figure(figsize=(12, 8))
emd.plotting.plot_imfs(imfs, fig=fig, sample_rate=sample_rate)
plt.title("IMFs of .wav Audio")
plt.show()
```

```python
import numpy as np
from scipy.io import wavfile
import matplotlib.pyplot as plt

# 1. Import the pipeline from your core script
from core import HHSAPipeline

# 2. Load and slice the audio
# (Remember to slice away the silence to avoid the math domain error!)
sample_rate, audio_data = wavfile.read(PREDICTOR_audio)
if audio_data.ndim > 1:
    audio_data = audio_data[:, 0]

audio_data = audio_data.astype(float)
audio_data = audio_data / np.max(np.abs(audio_data))

# Grab a 0.5-second snippet starting at 5 seconds
start_idx = int(5.0 * sample_rate)
end_idx = int(5.5 * sample_rate)
audio_snippet = audio_data[start_idx:end_idx]

if np.std(audio_snippet) == 0:
    raise ValueError("Audio snippet is silent!")
```

```python
# 3. Initialize the pipeline
# It automatically defaults to decomposition="iceemdan" and frequency_method="hybrid"
pipeline = HHSAPipeline(
    sample_rate=sample_rate,
    max_imfs=10,
    ensemble_size=64, # Your hardware can handle this easily
    noise_width=0.2
)

# 4. Fit the audio data
print("Running HHSA with ICEEMDAN...")
result = pipeline.fit(audio_snippet) # Returns an HHSAResult object
```
