## Exercise 1: Extracting MNE-BIDS Information

```python
from collections import Counter
from pathlib import Path
import numpy as np
"""
try:
    import mne
except ModuleNotFoundError:
    mne = None
    raw = None
    print("MNE is not installed in this environment. Install it with: pip install mne")
"""

if mne is not None:
    fif_path = Path('path_to_tutorial_01_file.fif')
    if fif_path.exists():
        raw = mne.io.read_raw_fif(fif_path, preload=True)
    else:
        sfreq = 100.0
        times = np.arange(0, 10, 1 / sfreq)
        ch_names = ['Fz', 'Cz', 'Pz', 'Oz']
        data = 1e-6 * np.vstack([
            np.sin(2 * np.pi * freq * times) for freq in [6, 8, 10, 12]
        ])
        info = mne.create_info(ch_names=ch_names, sfreq=sfreq, ch_types='eeg')
        info['line_freq'] = 60
        raw = mne.io.RawArray(data, info)
        raw.info['bads'] = []
        print(f"Using synthetic RawArray because {fif_path} was not found.")

    channel_counts = Counter(raw.get_channel_types())
    print("Channel counts by type:", dict(channel_counts))
    print("Sampling frequency:", raw.info['sfreq'], "Hz")
    print("Highpass filter:", raw.info['highpass'], "Hz")
    print("Lowpass filter:", raw.info['lowpass'], "Hz")
    print("Line noise frequency:", raw.info['line_freq'], "Hz")
    print("Bad channels:", raw.info['bads'])
```

## Exercise 2: Evoked Responses and SVD Decomposition

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import mne # Make sure mne is imported here!

if mne is None:
    print("Skipping the MNE epochs/SVD example because MNE is not installed.")
else:
    epochs_path = Path('sub-s01_task-faceFO_eeg.fif')

    if epochs_path.exists():
        epochs = mne.read_epochs(epochs_path, preload=True)
    else:
        sfreq = raw.info['sfreq']
        events = np.column_stack([
            np.arange(0, 20) * int(sfreq / 2),
            np.zeros(20, dtype=int),
            np.ones(20, dtype=int),
        ])
        epochs = mne.EpochsArray(
            np.repeat(raw.get_data()[None, :, :50], len(events), axis=0),
            raw.info,
            events=events,
            tmin=0,
            event_id={'synthetic': 1},
        )
        print(f"Using synthetic epochs because {epochs_path} was not found.")

    # --- THE FIX: Apply a standard spatial montage ---
    # on_missing='ignore' ensures it won't crash if a synthetic channel name is weird
    epochs.set_montage('standard_1020', on_missing='ignore')
    # ------------------------------------------------

    evoked = epochs.average()
    
    # We can now plot the joint figure without crashing
    fig_joint = evoked.plot_joint(show=False)

    evoked_data = evoked.data
    U, S, V = np.linalg.svd(evoked_data, full_matrices=False)

    n_components = min(5, U.shape[1])
    fig, axes = plt.subplots(1, n_components, figsize=(3 * n_components, 3))
    axes = np.atleast_1d(axes)

    for i, ax in enumerate(axes):
        mne.viz.plot_topomap(U[:, i], evoked.info, axes=ax, show=False)
        ax.set_title(f'Component {i + 1}')

    plt.suptitle('Topographies of the First SVD Components', y=1.1)
    plt.show()
```
