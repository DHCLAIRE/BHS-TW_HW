## Exercise 1: Extracting MNE-BIDS Information

```python
import mne

# Replace with the actual file path from Tutorial 01
raw = mne.io.read_raw_fif('path_to_tutorial_01_file.fif', preload=True)

# 1. How many channels do you have for each type of sensor?
print("Channel counts by type:", raw.get_channel_types().count('eeg'), "EEG channels")
# (You can use collections.Counter(raw.get_channel_types()) for a full breakdown)

# 2. What is the sampling frequency?
print("Sampling frequency:", raw.info['sfreq'], "Hz")

# 3. Have the data been filtered?
print("Highpass filter:", raw.info['highpass'], "Hz")
print("Lowpass filter:", raw.info['lowpass'], "Hz")

# 4. What is the frequency of the line noise?
print("Line noise frequency:", raw.info['line_freq'], "Hz")

# 5. Is there any bad channel?
print("Bad channels:", raw.info['bads'])
```

## Exercise 2: Evoked Responses and SVD Decomposition

```python
import mne
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the epoched file
# Note: Ensure the .fif file is in your current working directory or provide the full path
epochs = mne.read_epochs('sub-s01_task-faceFO_eeg.fif', preload=True)

# 2. Create the evoked response and make a joint plot of the time-series
evoked = epochs.average()
fig_joint = evoked.plot_joint(show=True)

# 3. Create a variable with the array of the evoked data 
# The shape of evoked_data will be (n_channels, n_times)
evoked_data = evoked.data

# 4. Decompose the data using numpy.linalg.svd
# U: Spatial patterns (n_channels, n_channels)
# S: Singular values
# V: Temporal patterns (n_times, n_times)
U, S, V = np.linalg.svd(evoked_data, full_matrices=False)

# 5. Make a matplotlib figure with 1 row and 5 columns and plot the first 5 components
fig, axes = plt.subplots(1, 5, figsize=(15, 3))

for i in range(5):
    # Plot the i-th column of U (the spatial topography for that component)
    mne.viz.plot_topomap(U[:, i], evoked.info, axes=axes[i], show=False)
    axes[i].set_title(f'Component {i+1}')

plt.suptitle('Topographies of the First 5 SVD Components', y=1.1)
plt.show()
```
