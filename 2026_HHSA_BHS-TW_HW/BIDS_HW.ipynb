```python
from bids import BIDSLayout
import pandas as pd
```

```python
# Initialize the BIDS layout (replace with your actual path)
layout = BIDSLayout('./ds000228')

# Get all BOLD nii.gz files for subject pixar003
pixar003_bold = layout.get(subject='pixar003', suffix='bold', extension='nii.gz', return_type='filename')

print("BOLD files for sub-pixar003:")
for file in pixar003_bold:
    print(file)
```

```python
# Extract metadata for all BOLD files
bold_files = layout.get(suffix='bold', extension='nii.gz')

# Create a set to capture unique flip angles across the dataset
flip_angles = set()

for f in bold_files:
    metadata = layout.get_metadata(f.path)
    if 'FlipAngle' in metadata:
        flip_angles.add(metadata['FlipAngle'])

print(f"Unique Flip Angles in DS000228: {flip_angles}")
```

```python
# Initialize layout for the Midnight Brain Scan dataset
layout_mbs = BIDSLayout('./midnight_brain_scan_dataset_folder')

# Locate the participants.tsv file
participants_file = layout_mbs.get(suffix='participants', extension='tsv', return_type='filename')[0]

# Load it into a pandas DataFrame
df_participants = pd.read_csv(participants_file, sep='\t')

# Display the first few rows
print(df_participants.head())
```
