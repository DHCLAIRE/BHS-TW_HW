### 1. Research question suited for deep learning

* **Example Question:** Can we accurately decode complex, continuous natural speech or semantic concepts from high-dimensional, non-stationary neural signals (such as multi-channel EEG or MEG data) as a participant listens to a narrative?
* **Deep Learning Advantage:** Traditional linear models (like standard regression) struggle with the immense non-linearity, high dimensionality, and complex spatiotemporal dynamics of non-invasive neural data. Deep learning architectures (such as Recurrent Neural Networks, Transformers, or 1D Convolutional Neural Networks) are exceptionally good at hierarchical feature extraction. They can automatically discover complex, non-linear interactions across time and electrode channels without requiring manual feature engineering.

### 2. Research question NOT suited for deep learning

* **Example Question:** Does increasing the semantic complexity of a sentence cause a statistically significant increase in the peak amplitude or latency of the N400 component in EEG tracking?
* **Deep Learning Disadvantage:** Deep learning models are inherently "black boxes" designed for high-accuracy prediction and classification rather than statistical inference or hypothesis testing. If the goal is to isolate a specific, well-defined biological or cognitive variable and establish a clear causal or parametric relationship, classical statistical methods (like Linear Mixed-Effects Models or ANOVA) are far superior. They offer clear p-values, interpretable effect sizes, and straightforward parameter estimates that deep networks lack.
