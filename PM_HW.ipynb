### 1. License for Sharing Code

**Which license do you pick and why?**

I would choose the **MIT License** because it is a free software license. It lets others use, modify, share, and sell your code with few restrictions. The main condition is that copies or substantial portions must include the original copyright notice and license text. This preserves the copyright of the original author and gives ownership of authors.

### 2. License for Sharing Data

**Which license do you pick and why?**

I would choose the **Creative Commons Attribution-NonCommercial 4.0 International (CC BY-NC 4.0)** license.

* **Why:** This license explicitly covers every requirement.
* **Attribution (BY):** Anyone who uses the data must give appropriate credit, provide a link to the license, and indicate if changes were made.
* **NonCommercial (NC):** Users are strictlyf prohibited from using the material for commercial purposes.
* **Modifications Allowed:** Because there is no "NoDerivatives" (ND) clause attached, other researchers are free to remix, adapt, and build upon your data for their own non-commercial research.

### 4. Neuroimaging Paper on an Open Science Framework

A prime example of modern open science practices in neuroimaging can be seen in papers that publish their datasets to the OpenNeuro archive and manage their workflows on platforms like GitHub or the Open Science Framework (OSF).

**1. Code Available?**

* **Availability:** Yes, the code is typically hosted on a public version-control repository like GitHub and linked directly to the OSF project page.
* **Standards Followed:** Researchers increasingly use version control (Git) to track changes and utilize software containers (like Docker or Singularity) to freeze the computing environment. This ensures that the exact software versions used for the analysis are preserved, making the code fully reproducible (Gorgolewski & Poldrack, 2016).

**2. Documentation for Data Analysis Available?**

* **Availability:** Yes, documentation is provided alongside the code and data.
* **Standards Followed:** Documentation is usually written in Markdown (via README files or OSF Wikis). For the actual data analysis, researchers often use "literate programming" environments like interactive Jupyter Notebooks. These notebooks weave together narrative text, analysis code, and visualizations into a single, executable document that walks the reader step-by-step through the methodology.

**3. Data Available?**

* **Availability:** Yes, raw and preprocessed data are hosted on field-specific repositories like OpenNeuro.
* **Standards Followed:** The primary standard followed is the **Brain Imaging Data Structure (BIDS)**. BIDS is a community standard that organizes neuroimaging data into a specific, easy-to-understand folder structure using simple file formats: NIfTI for brain images, JSON for metadata, and TSV (tab-separated values) for behavioral data (Gorgolewski & Poldrack, 2016). By adhering to BIDS, the data can be automatically ingested and processed by standardized analysis pipelines (like fMRIPrep) without manual reconfiguration.
